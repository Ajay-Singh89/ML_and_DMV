```python

# Experiment No. 1: Feature Transformation using PCA
# Aim: To reduce the dimensions of the Wine dataset using PCA
# and visualize the data using two principal components.


# STEP 1: Import required libraries

import pandas as pd                  # Used to load and manipulate datasets
import numpy as np                   # Used for numerical operations
import matplotlib.pyplot as plt      # Used to create graphs
import seaborn as sns                # Used for attractive data visualization

from sklearn.preprocessing import StandardScaler  # Used to standardize data
from sklearn.decomposition import PCA              # Used to perform PCA


# STEP 2: Load the Wine dataset

# Read the dataset directly from the given URL
url = "https://media.geeksforgeeks.org/wp-content/uploads/Wine.csv"

df = pd.read_csv(url)

# Display the first 5 rows of the dataset
print("First five rows:")
print(df.head())

# Display the number of rows and columns
print("\nDataset shape:", df.shape)

# Display all column names
print("\nDataset columns:")
print(df.columns.tolist())


# STEP 3: Separate features and target column

# Some datasets contain a column identifying the wine type.
# We search for a column with a commonly used target name.

target_candidates = ["type", "color", "wine_type", "class", "label"]

target_col = next(
    (col for col in df.columns if col.lower() in target_candidates),
    None
)

# If a target column exists, separate it from the features.
if target_col:
    y = df[target_col]                # Stores the wine labels
    X = df.drop(columns=[target_col]) # Stores the input features
else:
    y = None
    X = df.copy()                     # Uses all columns as features

# Select only numerical columns because PCA works with numerical data.
X = X.select_dtypes(include=[np.number])

# Replace missing values with the mean of their respective columns.
X = X.fillna(X.mean())


# STEP 4: Standardize the dataset

# PCA is affected by differences in feature scales.
# StandardScaler converts each feature to approximately:
# Mean = 0 and Standard Deviation = 1.

scaler = StandardScaler()

# Fit the scaler and transform the original data.
X_scaled = scaler.fit_transform(X)


# STEP 5: Apply PCA

# Create a PCA object with 2 principal components.
# This reduces the original features to just 2 dimensions.

pca = PCA(n_components=2)

# Transform the standardized data into two principal components.
X_pca = pca.fit_transform(X_scaled)


# STEP 6: Create a new DataFrame

# Store the transformed data in a DataFrame.
# PC1 and PC2 represent the two new features.

pca_df = pd.DataFrame(
    X_pca,
    columns=["Principal Component 1", "Principal Component 2"]
)

# Add the wine type column if it exists.
if y is not None:
    pca_df["Wine Type"] = y.astype(str).values

# Display the transformed dataset.
print("\nPCA Transformed Dataset:")
print(pca_df.head())


# STEP 7: Calculate explained variance

# Explained variance tells us how much information
# each principal component retains from the original data.

print("\nExplained Variance Ratio:")
print(pca.explained_variance_ratio_)

# Calculate the total percentage of variance retained
# by the two principal components.

total_variance = sum(pca.explained_variance_ratio_) * 100

print("\nTotal Variance Retained:", round(total_variance, 2), "%")


# STEP 8: Visualize the PCA results

# Set the size of the graph.
plt.figure(figsize=(9, 6))

# If wine labels are available, use different colors for each type.
if y is not None:

    sns.scatterplot(
        data=pca_df,
        x="Principal Component 1",
        y="Principal Component 2",
        hue="Wine Type",
        palette="Set1",
        s=60
    )

else:
    # If no labels exist, display all points in one scatter plot.
    plt.scatter(
        pca_df["Principal Component 1"],
        pca_df["Principal Component 2"],
        s=60
    )

# Add graph title and axis labels.
plt.title("PCA - Wine Dataset")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")

# Display grid lines.
plt.grid(True)

# Adjust spacing to prevent overlapping labels.
plt.tight_layout()

# Display the final graph.
plt.show()
```
